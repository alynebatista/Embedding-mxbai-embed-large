```python
from langchain_community.embeddings import OllamaEmbeddings

# Define o modelo
embeddings = OllamaEmbeddings(model="mxbai-embed-large")

# Define a frase a ser usada
query = "Bonito é melhor que feio. Explícito é melhor que implícito."

# Calcula os embeddings
query_embedding = embeddings.embed_query(query)

# Exibe os Embeddings
print(query_embedding)

# Calcula e exibe o tamanho do embedding
embedding_size = len(query_embedding)
print(f"Tamanho dos embeddings: {embedding_size}")
```
